# Symbolic Collapse Pressure Simulator

This interactive tool calculates symbolic collapse pressure using the core formula:

$$
\mathcal{C} = \frac{I \cdot S \cdot P}{T \cdot E}
$$

You can adjust the variables below to simulate collapse conditions for individuals, groups, or societies.

```
import ipywidgets as widgets
from IPython.display import display, Markdown

# Define sliders for each variable
I = widgets.FloatSlider(description='Information (I)', min=0.1, max=10000, step=10, value=3000)
S = widgets.FloatSlider(description='Abstraction (S)', min=0.1, max=10, step=0.1, value=5)
P = widgets.FloatSlider(description='Polarization (P)', min=0.0, max=1.0, step=0.01, value=0.5)
T = widgets.FloatSlider(description='Fidelity (T)', min=0.01, max=1.0, step=0.01, value=0.6)
E = widgets.FloatSlider(description='Coherence (E)', min=0.01, max=1.0, step=0.01, value=0.7)

def compute_collapse(I, S, P, T, E):
    C = (I * S * P) / (T * E)
    Psi = T * E * 1000  # Example adaptive threshold
    status = '⚠️ Collapse Likely' if C >= Psi else '✅ Stable'
    display(Markdown(f"### Collapse Pressure: {C:.2f}  "))
    display(Markdown(f"### Threshold: {Psi:.2f}  "))
    display(Markdown(f"## **{status}**"))

ui = widgets.VBox([I, S, P, T, E])
out = widgets.interactive_output(compute_collapse, {'I': I, 'S': S, 'P': P, 'T': T, 'E': E})
display(ui, out)
```
